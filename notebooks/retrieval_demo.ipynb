# Retrieval Demo

Runs 5 example queries against the arXiv cs.CL FAISS index (50 papers, 914 reference-stripped chunks) and shows the top-3 retrieved passages for each, using the shared `SearchEngine` from `scripts/search_lib.py`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "scripts"))
from search_lib import SearchEngine

engine = SearchEngine()
print(f"Index loaded: {engine.index.ntotal} chunks")
```

## Query 1

**How can hidden states be used to detect hallucinations in large language models?**

```python
query = 'How can hidden states be used to detect hallucinations in large language models?'
results = engine.search(query, k=3)
for r in results:
    print(f"[{r['rank']}] {r['chunk_id']}  (L2 distance={r['distance']:.4f})")
    print(r['text'][:400].replace(chr(10), ' '))
    print()
```
